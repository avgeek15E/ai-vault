# Sets

* defined by {}
* sets are unordered
* do not allow duplicates
* add
* remove and discard can be used for same purpose
* discard does not raise keyword error if the element is not present in the set
* pop removes element from the set from start
* clear is use to empty the set

```python
my_set = {1, 2, 3, 3}
print(my_set)
```

```python
my_set.add(4)
print(my_set)
```

```python
my_set.pop()
print(my_set)
```

```python
number = {1,2,3,4,5,6}
print(3 in my_set)
print(11 in my_set)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

union_set = set1.union(set2)
print(union_set)
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

print(set1.difference(set2))
print(set2.difference(set1))
```

```python
print(set1.symmetric_difference(set2))
```

```python
set1 = {1,2,3,4}
set2 = {1,2}

print(set1.issubset(set2))
print(set2.issubset(set1))
```

```python
set1 = {1,2,3,4}
set2 = (1,2)
print(set1.issubset(set2))
```
