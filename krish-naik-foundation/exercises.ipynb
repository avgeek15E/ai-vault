```python
def countdown(n):
    if n==0:
        print("Done")
    else:
        print(n)
        countdown(n-1)

countdown(5)
```

```python
def facto(n):
    if n==1:
        return 1
    else:
        return n * facto(n-1)

print(facto(5))
```

```python
def fibo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibo(n-2) + fibo(n-1)

print(fibo(11))
```

```python
year = int(input("Enter year"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print("Leap Year")
        else:
            print("Not leap")
    else:
        print("Leap Year")
else:
    print("Not leap")
```
